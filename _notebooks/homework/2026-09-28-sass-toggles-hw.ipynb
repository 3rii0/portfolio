3. The switch class controls the toggle's position, size, padding, and **alignment**.
4. The **border-radius** property makes the slider have rounded edges.
5. The `&:before` class dictates the **white circle** of the toggle.

// Comment 2: Selecting each toggle input and target element by ID allows JavaScript
// to link user input directly to page features (AP CSP Selection concept).
const themeToggle = outputElement.querySelector("#toggle-theme");
const badgeToggle = outputElement.querySelector("#toggle-badge");
const glowToggle = outputElement.querySelector("#toggle-glow");
const allToggles = outputElement.querySelectorAll(".ocs_toggle-input");

const previewCard = outputElement.querySelector("#preview-card");
const previewTitle = outputElement.querySelector("#preview-title");
const creatorBadge = outputElement.querySelector("#creator-badge");
const focusStatus = outputElement.querySelector("#focus-status");
const activeCountOutput = outputElement.querySelector("#active-count");

// Comment 3: I thought .checked was really useful because it returns a boolean (true or false),
// which lets us directly toggle CSS classes or set .hidden = !badgeToggle.checked cleanly.
function updateWorkspace() {
  // Feature 1: Switch between Light Mode and Midnight Dark Mode
  if (themeToggle.checked) {
    previewCard.classList.add("dark-theme");
    previewTitle.textContent = "Studio Workspace: Midnight Mode 🌙";
  } else {
    previewCard.classList.remove("dark-theme");
    previewTitle.textContent = "Studio Workspace: Light Mode ☀️";
  }

  // Feature 2: Show or hide the Verified Creator Badge based on Toggle 2's checked state
  creatorBadge.hidden = !badgeToggle.checked;

  // Feature 3: Toggle the neon pink focus glow border and status text
  if (glowToggle.checked) {
    previewCard.classList.add("glow-active");
    focusStatus.textContent = "Focus Mode: ON (Do Not Disturb Active ✨)";
  } else {
    previewCard.classList.remove("glow-active");
    focusStatus.textContent = "Focus Mode: OFF (Notifications allowed)";
  }

  // Comment 4: Looping over all .ocs_toggle-input checkboxes with forEach lets us count
  // exactly how many toggles are currently turned on every time a state changes.
  let activeCount = 0;
  allToggles.forEach(function(toggle) {
    if (toggle.checked) {
      activeCount += 1;
    }
  });

  activeCountOutput.textContent = activeCount + " / " + allToggles.length + " Toggles ON";
}

// Comment 5: Adding a 'change' event listener to every toggle makes the counter and preview
// update immediately without needing a separate submit button.
allToggles.forEach(function(toggle) {
  toggle.addEventListener("change", updateWorkspace);
});
