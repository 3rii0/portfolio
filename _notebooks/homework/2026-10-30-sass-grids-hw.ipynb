## Homework Hack

{% capture challenge8 %}
Homework Hack: fix both grids using only OCS classes. Press Run and keep going until every check says PASS.
{% endcapture %}

{% capture code8 %}
// Fix the HTML inside the quotes, then press Run.
let myHTML = [
  '<div class="ocs_grid ocs_grid--standard">',
  '  <div class="ocs_grid-cell ocs_grid-cell--header">Temperature Trial (°C)</div>',
  '  <div class="ocs_grid-cell">Beaker 1: 22.4</div>',
  '  <div class="ocs_grid-cell">Beaker 2: 22.9</div>',
  '  <div class="ocs_grid-cell ocs_grid-cell--accent">Beaker 3: 31.6</div>',
  '</div>',
  '<div class="ocs_grid ocs_grid--calculator">',
  '  <div class="ocs_grid-cell">7</div>',
  '  <div class="ocs_grid-cell">8</div>',
  '  <div class="ocs_grid-cell">9</div>',
  '  <div class="ocs_grid-cell ocs_grid-cell--accent">Avg</div>',
  '</div>'
].join("");

// --- Validation checks ---
let classList = [];
for (let found of (myHTML.match(/class="[^"]*"/g) || [])) {
  classList = classList.concat(found.slice(7, -1).split(' ').filter(c => c !== ''));
}
let madeUp = classList.filter(c => !c.startsWith('ocs_') && !c.startsWith('cols-'));

function line(label, ok) { return '<li>' + (ok ? 'PASS ' : 'FIX ') + label + '</li>'; }
let checks = '<ul>' +
  line('No inline style attributes', !myHTML.includes('style=')) +
  line('No made up classes' + (madeUp.length > 0 ? ' (found: ' + madeUp.join(', ') + ')' : ''), madeUp.length === 0) +
  line('Has an ocs_grid root', classList.includes('ocs_grid')) +
  line('Uses the standard variant for Grid 1', classList.includes('ocs_grid--standard')) +
  line('Uses the calculator variant for Grid 2', classList.includes('ocs_grid--calculator')) +
  line('Every box uses ocs_grid-cell', classList.filter(c => c === 'ocs_grid-cell').length >= 8) +
  line('Title is a header cell', classList.includes('ocs_grid-cell--header')) +
  line('Has accent cells (outlier beaker and Avg key)', classList.filter(c => c === 'ocs_grid-cell--accent').length >= 2) +
'</ul>';

outputElement.innerHTML = checks + myHTML;
{% endcapture %}
