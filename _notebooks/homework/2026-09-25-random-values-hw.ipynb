## Popcorn Hack 1: Random SFI Record Number

**Prediction & Explanation:**
- **Possible Range:** Using `random.randint(1, record_count)` with `record_count = 4` can return `1`, `2`, `3`, or `4` because `randint(a, b)` is **inclusive** of both endpoints.
- **Why Repeats Are Allowed:** Each call to `random.randint(1, 4)` is an independent random selection rather than a guaranteed cycle/rotation, so the same record ID can appear twice or more in a row.
- **Modified Input:** When `record_count` is changed to `6`, the complete possible output set expands to `1, 2, 3, 4, 5, 6`.

```python
# Popcorn Hack 1 - Random SFI Record Number

import random

record_count = 4

print(f"--- Run 1: Selecting from {record_count} SFI Records (Possible IDs: 1 to {record_count} inclusive) ---")
for trial in range(1, 6):
    selected_id = random.randint(1, record_count)
    print(f"  Trial {trial}: Selected SFI Record ID = {selected_id}")

print("Explanation: Both endpoints (1 and 4) are reachable, and repeats occur naturally because each draw is independent.")

# Change one input (increase record_count to 6)
record_count_modified = 6
print(f"\n--- Run 2: Modified Input (record_count = {record_count_modified}, Possible IDs: 1 to {record_count_modified} inclusive) ---")
for trial in range(1, 6):
    selected_id = random.randint(1, record_count_modified)
    print(f"  Trial {trial}: Selected SFI Record ID = {selected_id}")
```

## Popcorn Hack 3: Random SFI API Test

**Prediction & Explanation:**
- **Predicted Action Outcomes:**
  1. If `selected_action == "GET/search"`, the script routes to the search branch and prints a lookup message for the chosen part's `spec_number`.
  2. If `selected_action == "POST/create"`, the script routes to the creation branch and validates the chosen part before inserting it.
- **Modified Action List:** Adding `"DELETE/remove"` increases the total possible `(part, action)` combinations from $3 \times 3 = 9$ outcomes to $3 \times 4 = 12$ possible outcomes.

```python
# Popcorn Hack 3 - Random SFI API Test

import random

parts = [
    {"product_name": "Replacement Flywheels", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "spec_number": "2.1"}
]

actions = [
    "GET/search",
    "POST/create",
    "PUT/update"
]

def run_api_test(part_list, action_list):
    chosen_part = random.choice(part_list)
    chosen_action = random.choice(action_list)

    print(f"Selected Part: {chosen_part['product_name']} (Spec {chosen_part['spec_number']})")
    print(f"Selected Action: {chosen_action}")

    if chosen_action == "GET/search":
        print(f"  -> Backend Response: 200 OK - Returning record for Spec {chosen_part['spec_number']}.\n")
    elif chosen_action == "POST/create":
        print(f"  -> Backend Response: 201 Created - Validating and storing '{chosen_part['product_name']}'.\n")
    elif chosen_action == "PUT/update":
        print(f"  -> Backend Response: 200 OK - Updating metadata for Spec {chosen_part['spec_number']}.\n")
    else:
        print(f"  -> Backend Response: 200 OK - Removing Spec {chosen_part['spec_number']} from active records.\n")

print("--- Run 1: Standard 3 Actions (GET, POST, PUT) ---")
run_api_test(parts, actions)

# Change action list to include DELETE/remove and run again
actions_modified = actions + ["DELETE/remove"]
print("--- Run 2: Modified Action List (+ DELETE/remove) ---")
run_api_test(parts, actions_modified)
```

## Homework Hack (Independent): SFI Backend QA Simulator

**Simulator Logic & Complete Output Space Explanation:**
- **Possible Output Set:** There are `3` structured SFI part records in `parts` and `4` API actions in `actions`, resulting in $3 \times 4 = 12$ possible `(part, action)` combinations on any single run.
- **Endpoints & Repeats:** Every part and every action is reachable on each run. Because each iteration makes an independent random selection using `random.choice()`, repeated parts or actions across consecutive runs are normal and expected.
- **Duplicate-Spec Handling:** When `"POST/create"` is selected, the simulator checks whether `selected_part["spec_number"] in existing_spec_numbers` (`["1.1", "2.1"]`). Specs `"1.1"` and `"2.1"` are rejected as duplicates, while Spec `"1.2"` is accepted as a new spec.

```python
# Homework Hack - SFI Backend QA Simulator

import random

parts = [
    {
        "product_name": "Replacement Flywheels",
        "category": "Auto Racing",
        "spec_number": "1.1"
    },
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Drag Racing",
        "spec_number": "1.2"
    },
    {
        "product_name": "Racing Flywheel Record",
        "category": "Auto Racing",
        "spec_number": "2.1"
    }
]

actions = [
    "GET/search",
    "POST/create",
    "PUT/update",
    "DELETE/remove"
]

existing_spec_numbers = ["1.1", "2.1"]
test_count = 5

def simulate_sfi_backend_qa(part_records, api_actions, existing_specs, runs):
    total_combinations = len(part_records) * len(api_actions)
    print(f"=== SFI Backend QA Simulator ({runs} Runs | {total_combinations} Possible Part/Action Pairs) ===\n")

    for run_num in range(1, runs + 1):
        selected_part = random.choice(part_records)
        selected_action = random.choice(api_actions)

        name = selected_part["product_name"]
        category = selected_part["category"]
        spec = selected_part["spec_number"]
        is_duplicate = spec in existing_specs

        print(f"Run {run_num}:")
        print(f"  Selected Record -> Name: '{name}' | Category: '{category}' | Spec: '{spec}'")
        print(f"  Selected Action -> {selected_action}")

        if selected_action == "GET/search":
            print(f"  Backend Result  -> 200 OK: Found '{name}' (Spec {spec}) in {category} catalog.\n")
        elif selected_action == "POST/create":
            if not is_duplicate:
                print(f"  Backend Result  -> 201 CREATED: Spec {spec} is unique. Added '{name}' to backend.\n")
            else:
                print(f"  Backend Result  -> 409 CONFLICT (REJECTED): Duplicate Spec {spec} already exists in {existing_specs}.\n")
        elif selected_action == "PUT/update":
            print(f"  Backend Result  -> 200 OK: Updated effective metadata for '{name}' (Spec {spec}).\n")
        elif selected_action == "DELETE/remove":
            print(f"  Backend Result  -> 200 OK: Archived and removed Spec {spec} ('{name}') from active list.\n")

# Run the simulator for the required test_count (5 runs)
simulate_sfi_backend_qa(parts, actions, existing_spec_numbers, test_count)
```
